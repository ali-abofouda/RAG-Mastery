```python
# directory loader is used to load multiple files from a directory. It can be used to load text files, PDFs, and other document types. The directory loader will recursively search through the specified directory and load all files that match the specified file types.
from langchain_community.document_loaders import DirectoryLoader


directory_loader = DirectoryLoader("data", glob="**/*.txt", loader_cls=TextLoader, loader_kwargs={"encoding": "utf-8"} , show_progress=True)
documents = directory_loader.load()

print("Loaded Documents:")
for doc in documents:
    print(doc.page_content[:100] + "...")
    print("-" * 40)
    print("Metadata:")
    print(doc.metadata)
    print("=" * 80)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter, CharacterTextSplitter, TokenTextSplitter
```

```python
## method 1: Using character-based text splitting

text = documents[0].page_content  # Use the content of the first loaded document
text    
```

```python
print("Character Text Splitter")
character_splitter = CharacterTextSplitter(
    separator="\n",  # Split on new lines
    chunk_size=500,   # Maximum number of characters in each chunk
    chunk_overlap=50  # Number of characters to overlap between chunks
)
character_chunks = character_splitter.split_text(text)
print(f"Number of chunks created: {len(character_chunks)}")
print("*"*40)
print("Sample chunk:")
print(character_chunks[0])  # Print the first chunk
print("/"*80)
print("second chunk:")
print(character_chunks[1])  # Print the second chunk
print("/"*80)
print("third chunk:")
print(character_chunks[2])  # Print the third chunk
print("/"*80)
print("section metadata:")
print(documents[0].metadata)  # Print the metadata of the first document
```

```python
## second method: Using recursive character-based text splitting
print("Recursive Character Text Splitter")
recursive_splitter = RecursiveCharacterTextSplitter(
    separators=["\n\n", "\n", " ", ""],  # Try to split on paragraphs, then lines, then spaces
    chunk_size=500,   # Maximum number of characters in each chunk
    chunk_overlap=50  # Number of characters to overlap between chunks
)
recursive_chunks = recursive_splitter.split_text(text)
print(f"Number of chunks created: {len(recursive_chunks)}")
print("-"*80)
print("first chunk:")
print("-"*20)
print(f"{recursive_chunks[0]}")
print("-"*40)
print("second chunk:")
print("-"*20)
print(recursive_chunks[1])  # Print the second chunk
print("-"*40)
print("third chunk:")
print("-"*20)
print(recursive_chunks[2])  # Print the third chunk
print("-"*40)
print("section metadata:")
print(documents[0].metadata)  # Print the metadata of the first document
```
