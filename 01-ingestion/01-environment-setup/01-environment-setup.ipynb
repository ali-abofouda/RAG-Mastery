### الخطوة 2: اختبار استيراد المكتبات الأساسية لمنظومة LangChain والـ Ingestion

المكتبات المطلوبة:
* `langchain`: إطار العمل الأساسي
* `langchain_community`: أدوات التكامل والـ Document Loaders
* `pypdf`: قراءة واستخراج ملفات PDF
* `faiss`: قاعدة البيانات المتجهة المحلية الفائقة
* `dotenv`: تحميل مفاتيح الـ API السرية

```python
try:
    import langchain
    import langchain_community
    import pypdf
    import faiss
    import dotenv
    
    print(f"✅ langchain version: {langchain.__version__}")
    print(f"✅ langchain_community loaded successfully!")
    print(f"✅ pypdf version: {pypdf.__version__}")
    print(f"✅ faiss-cpu is ready!")
    print(f"✅ python-dotenv is available!")
except ImportError as e:
    print(f"⚠️ Missing package: {e}")
    print("To install dependencies using UV, run: uv add -r requirements.txt")
```

### الخطوة 3: تجهيز وقراءة المتغيرات البيئية ومفاتيح الـ API (`.env`)

نستخدم `python-dotenv` لتحميل مفاتيح الـ API (مثل `OPENAI_API_KEY` أو `GROQ_API_KEY`) بأمان.

```python
import os
from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv()

has_openai = "OPENAI_API_KEY" in os.environ
has_groq = "GROQ_API_KEY" in os.environ

print(f"OpenAI API Key configured: {has_openai}")
print(f"Groq API Key configured: {has_groq}")
```
