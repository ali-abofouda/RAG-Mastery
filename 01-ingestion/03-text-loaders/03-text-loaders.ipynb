# 03 - قراءة واستيراد الملفات النصية (Text Loaders & Directory Loaders)

في هذا الدفتر نطبق عملياً كيفية استيراد الملفات النصية الخام (`.txt`) في LangChain:
1. **`TextLoader`**: لتحميل ملف نصي منفرد ومحدد بدقة.
2. **`DirectoryLoader`**: لتحميل مجلد كامل يحوي عدة ملفات نصية دفعة واحدة باستخدام أنماط البحث (Glob Patterns).

### الخطوة 1: إنشاء مجلد وملفات نصية تجريبية برمجياً

سنقوم بإنشاء مجلد `data/text_files/` وكتابة ملفين تجريبيين:
* `python_intro.txt`: مقدمة عن لغة بايثون.
* `ml_basics.txt`: مقدمة عن أساسيات تعلم الآلة.

```python
import os

# إنشاء مجلد البيانات
data_dir = os.path.join("data", "text_files")
os.makedirs(data_dir, exist_ok=True)

# نصوص العينة
sample_files = {
    os.path.join(data_dir, "python_intro.txt"): (
        "Python is a versatile high-level programming language known for its readability and simplicity. "
        "It is widely used in artificial intelligence, web development, data science, and automation. "
        "Python's extensive ecosystem of libraries like NumPy, Pandas, and LangChain makes it the leading choice for AI applications."
    ),
    os.path.join(data_dir, "ml_basics.txt"): (
        "Machine learning is a branch of artificial intelligence focused on building applications that learn from data "
        "and improve their accuracy over time without being explicitly programmed. "
        "Key paradigms include supervised learning, unsupervised learning, and reinforcement learning."
    )
}

# كتابة الملفات بترميز UTF-8
for path, content in sample_files.items():
    with open(path, "w", encoding="utf-8") as f:
        f.write(content)

print(f"✅ تم إنشاء {len(sample_files)} ملفات نصية بنجاح داخل: {data_dir}")
```

### الخطوة 2: قراءة ملف نصي منفرد باستخدام `TextLoader`

نمرر مسار الملف ومعامل الترميز `encoding='utf-8'` لضمان قراءة كافة الحروف بشكل سليم.

```python
from langchain_community.document_loaders import TextLoader

# تهيئة المحمل لملف محدد
single_file_path = os.path.join(data_dir, "python_intro.txt")
loader = TextLoader(single_file_path, encoding="utf-8")

# تحميل الوثيقة
documents = loader.load()

print(f"نوع المخرج: {type(documents)}")
print(f"عدد الوثائق: {len(documents)}")
```

### الخطوة 3: فحص محتوى الوثيقة والميتا-داتا المستخرجة تلقائياً

```python
doc = documents[0]

print("--- البيانات الوصفية (Metadata) ---")
print(doc.metadata)
print(f"المصدر المستخرج: {doc.metadata.get('source')}")

print("\n--- أول 100 حرف من المحتوى النصي ---")
print(f"{doc.page_content[:100]}...")
```

### الخطوة 4: تحميل مجلد كامل دفعة واحدة باستخدام `DirectoryLoader`

المعاملات المستخدمة:
* `path`: مسار المجلد.
* `glob`: نمط المطابقة لجميع ملفات الـ `.txt` (`**/*.txt`).
* `loader_cls`: استخدام `TextLoader` لقراءة كل ملف.
* `loader_kwargs`: تمرير `encoding='utf-8'` لكل ملف.
* `show_progress`: عرض شريط تقدم العملية.

```python
from langchain_community.document_loaders import DirectoryLoader

batch_loader = DirectoryLoader(
    path=data_dir,
    glob="**/*.txt",
    loader_cls=TextLoader,
    loader_kwargs={"encoding": "utf-8"},
    show_progress=True
)

all_docs = batch_loader.load()
print(f"\n✅ إجمالي الوثائق المحملة من المجلد: {len(all_docs)}")
```

### الخطوة 5: استعراض جميع الوثائق المستخرجة من المجلد

```python
for idx, d in enumerate(all_docs, 1):
    print(f"الوثيقة {idx}:")
    print(f"  - المصدر: {d.metadata.get('source')}")
    print(f"  - عدد الأحرف: {len(d.page_content)}")
    print(f"  - مقتطف: {d.page_content[:80]}...")
    print("-" * 40)
```

### خلاصة الدرس والانتقال للمرحلة التالية:

تعلمنا في هذا الدرس:
1. استخدام `TextLoader` للملفات الفردية مع ضبط ترميز `UTF-8`.
2. استخدام `DirectoryLoader` مع أنماط المطابقة `glob` ومحدد الفئة `loader_cls=TextLoader` لتحميل مجلدات كاملة.
