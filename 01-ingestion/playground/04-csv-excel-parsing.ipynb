# مساحة تجارب CSV & Excel Tabular Data (Playground)

> اكتب هنا كودك وتجاربك الحرة الخاصة بقراءة وتحليل ملفات الجداول (CSV و Excel).
> البيانات التجريبية متوفرة في `data/csv/products.csv` و `data/excel/inventory.xlsx`.

```python
# 1. استيراد المكتبات وفحص المسارات
from pathlib import Path
import pandas as pd
from langchain_community.document_loaders import CSVLoader

csv_path = Path("data/csv/products.csv")
excel_path = Path("data/excel/inventory.xlsx")

print(f"CSV Exists: {csv_path.exists()}")
print(f"Excel Exists: {excel_path.exists()}")
```

```python
# 2. ابدأ تجاربك الحرة هنا

df = pd.read_csv(csv_path)
df
```

```python
from langchain_core.documents import Document

documents = []

for _, row in df.iterrows():
    content = f"""
Product: {row["product"]}
Description: {row["description"]}
""".strip()

    metadata = {
        "category": row["category"],
        "price": row["price"],
    }

    documents.append(
        Document(
            page_content=content,
            metadata=metadata
        )
    )
```

```python
print(f"Number of documents created: {len(documents)}")
print("-" * 40)
print("Sample document page content:")
print(documents[0].page_content)
print("-" * 40)
print("Sample document metadata:")
print(documents[0].metadata)
```
