# كشكول التجارب: قراءة بيانات الـ JSON المتداخلة

أهلاً بيك! ملفات الـ JSON مش دايماً بتكون مسطحة، ساعات كتير بيكون جواها كائنات متداخلة وقوائم (`Nested Objects`).

في الكشكول ده هنجرب:
1. قراءة ملف `company_data.json`.
2. تفكيك الأقسام والموظفين وتحويل كل موظف لمستند مستقل ينفع نبحث فيه.


```
# 1. استيراد المكتبات وتحميل ملف الـ JSON
from pathlib import Path

# نحدد مسار مجلد data تلقائياً
repo_root = Path.cwd()
while not (repo_root / "data").exists() and repo_root.parent != repo_root:
    repo_root = repo_root.parent
data_dir = repo_root / "data"

import json
from langchain_core.documents import Document

json_file = data_dir / "json" / "company_data.json"
with open(json_file, "r", encoding="utf-8") as f:
    data = json.load(f)

print("مفاتيح الـ JSON الأساسية:", list(data.keys()))
```

### 2. تفكيك البيانات المتداخلة لمستندات واضحة
تعال نلف على الأقسام والموظفين ونعمل لكل موظف كائن `Document` يحمل اسمه وقسمه ودوره:


```
docs_list = []

company_info = data.get("company", {})
departments = company_info.get("departments", [])

for dept in departments:
    dept_name = dept.get("name")
    for emp in dept.get("employees", []):
        emp_text = f"الموظف {emp.get('name')} يعمل في قسم {dept_name} بوظيفة {emp.get('role')}."
        docs_list.append(
            Document(
                page_content=emp_text,
                metadata={"emp_id": emp.get("id"), "department": dept_name}
            )
        )

print(f"تم إنشاء {len(docs_list)} مستند للموظفين!")
print("\nعينة من المستندات:")
for d in docs_list[:3]:
    print(f"- {d.page_content} (الميتا-داتا: {d.metadata})")
```
