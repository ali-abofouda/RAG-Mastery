# كشكول التجارب: قراءة ملفات مايكروسوفت وورد (DOCX)

أهلاً بيك! ملفات الوورد (`.docx`) ميزتها إنها بتحتوي على هيكل منظم: عناوين (`Headings`)، فقرات، وجداول.

في الكشكول ده هنجرب طريقتين:
1. طريقة سريعة ومباشرة باستخدام `Docx2txtLoader` لقراءة كل النصوص مرة واحدة.
2. طريقة دقيقة باستخدام مكتبة `python-docx` علشان نستخرج الجداول والعناوين بشكل منظم.


```
# 1. نستورد الأدوات ونفحص ملف الوورد
from pathlib import Path

# نحدد مسار مجلد data تلقائياً
repo_root = Path.cwd()
while not (repo_root / "data").exists() and repo_root.parent != repo_root:
    repo_root = repo_root.parent
data_dir = repo_root / "data"

from langchain_community.document_loaders import Docx2txtLoader
import docx

docx_file = data_dir / "docx" / "proposal.docx"
print(f"ملف الوورد: {docx_file.name} (موجود: {docx_file.exists()})")
```

### 2. الطريقة السريعة: Docx2txtLoader
دي بتديك النص كله في مستند واحد في ثواني:


```
loader = Docx2txtLoader(str(docx_file))
docs = loader.load()

print(f"طول النص بالكامل: {len(docs[0].page_content)} حرف")
print("\nأول جزء من المقترح:\n" + docs[0].page_content[:200] + "...")
```

### 3. الطريقة المنظمة: استخراج الجداول والعناوين
تعال نشوف إزاي نستخرج الجداول ونحولها لنص منظم ينفع الفيكتور يفهمه:


```
doc = docx.Document(str(docx_file))

print(f"عدد الفقرات في الملف: {len(doc.paragraphs)}")
print(f"عدد الجداول الموجودة: {len(doc.tables)}")

# نطبع محتوى أول جدول لو موجود
if doc.tables:
    table = doc.tables[0]
    print("\nمحتوى أول جدول في الملف:")
    for row in table.rows:
        row_text = [cell.text.strip() for cell in row.cells]
        print(" | ".join(row_text))
```
