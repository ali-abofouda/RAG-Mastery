# كشكول التجارب: قراءة واستخراج ملفات الـ PDF

أهلاً بيك! ملفات الـ PDF من أكتر أنواع الملفات استخداماً في الشركات، وفي نفس الوقت من أكترها مشكلات!

في الكشكول ده هنجرب:
1. محمل الـ PDF السريع والممتاز `PyMuPDFLoader`.
2. استخراج النص صفحة بصفحة مع الميتا-داتا ورقم كل صفحة.
3. تجربة دالة تنظيف خفيفة بتشيل المسافات الزيادة والأرقام العشوائية في الهوامش.


```
# 1. نستورد الأدوات ونحدد مسار ملف الـ PDF
from pathlib import Path

# نحدد مسار مجلد data تلقائياً
repo_root = Path.cwd()
while not (repo_root / "data").exists() and repo_root.parent != repo_root:
    repo_root = repo_root.parent
data_dir = repo_root / "data"

from langchain_community.document_loaders import PyMuPDFLoader

pdf_file = data_dir / "pdf" / "sample_financial_report.pdf"
if not pdf_file.exists():
    pdf_file = data_dir / "pdf" / "sample_paper.pdf"

print(f"الملف اللي هنجرب عليه: {pdf_file.name}")
```

### 2. قراءة صفحات الـ PDF
تعال نشغل `PyMuPDFLoader` ونشوف استخرج كام صفحة:


```
loader = PyMuPDFLoader(str(pdf_file))
pages = loader.load()

print(f"عدد الصفحات المستخرجة: {len(pages)} صفحة")
print(f"الميتا-داتا بتاعة أول صفحة: {pages[0].metadata}")
print("\nعينة من نص الصفحة الأولى:\n" + pages[0].page_content[:200] + "...")
```

### 3. تنظيف النصوص المستخرجة
ملفات الـ PDF ساعات بيكون فيها مسافات كتيرة أو كلمات مقطوعة بسبب نهاية السطر. تعال نعمل دالة تنظيف سريعة:


```
import re

def clean_pdf_text(raw_text: str) -> str:
    # نشيل المسافات المتكررة ونصلح الفواصل
    cleaned = re.sub(r'\s+', ' ', raw_text)
    return cleaned.strip()

sample_cleaned = clean_pdf_text(pages[0].page_content)
print("النص بعد التنظيف:")
print(sample_cleaned[:200] + "...")
```
