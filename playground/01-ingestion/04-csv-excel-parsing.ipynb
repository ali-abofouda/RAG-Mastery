# كشكول التجارب: قراءة الجداول وبيانات الإكسل (CSV & Excel)

أهلاً بيك! الجداول والـ Spreadsheets فيها ميزة مهمة: البيانات جواها عبارة عن أرقام وأعمدة.  
بس عشان نستخدمها في الـ RAG والبحث الدلالي، محتاجين نحول كل صف لجملة مفيدة يفهمها نموذج التضمين (`Embeddings`).

في الكشكول ده هنجرب:
1. قراءة ملف `products.csv` باستخدام `CSVLoader`.
2. قراءة ملف `inventory.xlsx` باستخدام `pandas`.
3. تحويل الصفوف لجمل سردية واضحة.


```
# 1. استيراد المكتبات وتحديد المسارات
from pathlib import Path

# نحدد مسار مجلد data تلقائياً
repo_root = Path.cwd()
while not (repo_root / "data").exists() and repo_root.parent != repo_root:
    repo_root = repo_root.parent
data_dir = repo_root / "data"

import pandas as pd
from langchain_community.document_loaders import CSVLoader

csv_file = data_dir / "tabular" / "products.csv"
excel_file = data_dir / "tabular" / "inventory.xlsx"

print(f"ملف الـ CSV: {csv_file.name}")
print(f"ملف الإكسل: {excel_file.name}")
```

### 2. قراءة ملف الـ CSV
`CSVLoader` بيعتبر كل صف عبارة عن `Document` مستقل:


```
loader = CSVLoader(str(csv_file), encoding="utf-8")
csv_docs = loader.load()

print(f"عدد المنتجات المستخرجة من الـ CSV: {len(csv_docs)}")
print("\nشكل أول منتج:")
print(csv_docs[0].page_content)
```

### 3. قراءة الإكسل وتحويل الصفوف لجمل مفيدة
تعال نقرأ الإكسل بـ `pandas` ونعمل جملة لطيفة لكل صف:


```
df = pd.read_excel(excel_file)
print(f"أعمدة الإكسل: {list(df.columns)}")
print(f"عدد الصفوف: {len(df)}\n")

print("أول 3 صفوف متحولين لجمل دلالية:")
for idx, row in df.head(3).iterrows():
    sentence = " | ".join([f"{col}: {val}" for col, val in row.items()])
    print(f"منتج {idx + 1}: {sentence}")
```
