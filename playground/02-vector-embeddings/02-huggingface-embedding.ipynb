```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ["HUGGINGFACEHUB_API_TOKEN"] = os.getenv("HUGGINGFACEHUB_API_TOKEN")
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

query_embedding = embeddings.embed_query("What is artificial intelligence?")

print(len(query_embedding))
print(query_embedding)
print(type(query_embedding))
```

```python
multiple_query_embeddings = embeddings.embed_documents(["What is artificial intelligence?", "What is machine learning?"])
print(len(multiple_query_embeddings))
print(multiple_query_embeddings)
```
