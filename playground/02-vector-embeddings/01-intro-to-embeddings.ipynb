أهلاً بيك! الكمبيوتر مبيفهمش الكلام اللي إحنا بنكتبه (زي "أنا بحب البرمجة" أو "الطقس مشمس اليوم"). الكمبيوتر بيفهم بس **أرقام**.

فكرة الـ **Embedding** ببساطة شديدة:
* بناخد الكلمة أو الجملة، ونعديها على نموذج ذكاء اصطناعي متخصص.
* النموذج بيحول الجملة دي لـ **سلسلة من الأرقام** (بنسميها Vector).
* الميزة الكبيرة: الجمل اللي معناها شبه بعض، بتتحول لأرقام قريبة من بعض في الفضاء الرياضي!

تعال نجرب نحول أول جملة لأرقام ونشوف شكلها عامل إزاي!


```python
# 1. نستورد موديل سريع ومجاني
from sentence_transformers import SentenceTransformer

# بنحمل موديل صغير ومشهور جداً اسمه all-MiniLM-L6-v2
model = SentenceTransformer("all-MiniLM-L6-v2")

print("الموديل جاهز لتحويل النصوص لمتجهات!")
```

### 2. تعال نحول جملة لأرقام
تعال نكتب جملة بسيطة ونشوف هتطلع كام رقم:


```python
text = "Artificial intelligence and machine learning"

# الموديل بيحول النص لمتجه أرقام
vector = model.encode(text)

print(f"الجملة الأصلية: '{text}'")
print(f"عدد الأبعاد (طول المتجه): {len(vector)} رقم!")
print(f"\nتعال نبص على أول 10 أرقام من المتجه:\n{vector[:10]}")
```

### 3. تجربة الفرق الدلالي بين كلمتين
تعال نقارن بين 3 جمل: اتنين بيتكلموا عن البرمجة، وواحدة بتتكلم عن البيتزا!


```python
from sklearn.metrics.pairwise import cosine_similarity

sent1 = "Python is great for data science"
sent2 = "Coding in Python is fun"
sent3 = "I ordered a pepperoni pizza for dinner"

v1 = model.encode([sent1])
v2 = model.encode([sent2])
v3 = model.encode([sent3])

print("التشابه بين (بايثون وبايثون):", round(float(cosine_similarity(v1, v2)[0][0]), 3))
print("التشابه بين (بايثون والبيتزا):", round(float(cosine_similarity(v1, v3)[0][0]), 3))
```
