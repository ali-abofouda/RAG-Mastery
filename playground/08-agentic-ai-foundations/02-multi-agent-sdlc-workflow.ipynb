# معمل محاكاة فريق تطوير البرمجيات الرشيقة (Agile Multi-Agent SDLC)

أهلاً بك في المعمل العملي للمحاضرة 56.

في هذا المعمل، هنشوف تطبيق واقعي جداً لأي شركة برمجيات شغالة بمنهجية Agile:
إزاي بنحول أدوار فريق العمل البشري إلى **فريق وكلاء ذكاء اصطناعي (Agentic AI Team)** بيتعاونوا مع بعض:
1. **وكيل إدارة المنتج (PM Agent)**: بياخد طلب العميل ويقسمه لمهمات وSprint Backlog.
2. **وكيل المطور (Developer Agent)**: بيكتب كود الحل البرمجي.
3. **وكيل مراجعة الكود (Code Reviewer Agent)**: بيراجع جودة الكود والأمان.
4. **وكيل فحص الجودة (QA Testing Agent)**: بيشغل الاختبارات ويكتشف الأخطاء.
5. **المشرف البشري في الحلقة (Human-in-the-Loop - HITL)**: بياخد القرار النهائي قبل النشر للإنتاج!


```python
# استيراد الأدوات وتجهيز البيئة
from typing import Dict, List, Any, Optional, Tuple

print('تم تجهيز معمل فريق تطوير البرمجيات بنجاح.')
```

### 1. وكيل إدارة المنتج (Product Manager Agent)

تعال نشوف إزاي الـ PM Agent بيستقبل رغبة العميل ويقسمها لمهمات محددة (User Stories).


```python
def pm_agent_decompose_sprint(client_request: str) -> List[Dict[str, str]]:
    print(f"طلب العميل الوارد: '{client_request}'")
    # تقسيم الطلب لقصص مستخدمين محددة
    sprint_tasks = [
        {"id": "TASK-1", "title": "محرك حساب نسبة نمو الإيرادات", "status": "جديد"},
        {"id": "TASK-2", "title": "فلتر تنقية وفحص المدخلات المالية", "status": "جديد"}
    ]
    return sprint_tasks

current_sprint = pm_agent_decompose_sprint("تطوير لوحة تحليلات مالية للشركة")
print(f"عدد المهمات المخططة للسباق: {len(current_sprint)}")
for t in current_sprint:
    print(f"  - [{t['id']}]: {t['title']}")
```

### 2. وكلاء التطوير والمراجعة والاختبار وحلقة التغذية الراجعة

لاحظ هنا إزاي المطور بيكتب كود أولي (قد يحتوي على خطأ زي القسمة على صفر).
وكيل الـ QA بيكتشف الخطأ ويبعت تقرير للمطور، فالمطور يصلحه ذاتياً قبل ما يروح للبشري!


```python
def dev_agent_write_code(task_id: str, feedback: Optional[str] = None) -> str:
    # لو في فيدباك من التست، المطور بيصلح الكود
    if feedback:
        print(f"[المطور]: جاري إصلاح الخطأ المكتشف ({feedback})...")
        return "def calculate_growth(q1, q3):\n    if q1 <= 0: raise ValueError('Baseline must be > 0')\n    return ((q3 - q1) / q1) * 100"
    else:
        print("[المطور]: كتابة النسخة الأولية من الكود...")
        return "def calculate_growth(q1, q3):\n    return ((q3 - q1) / q1) * 100"

def reviewer_agent_audit(code: str) -> Tuple[bool, str]:
    # مراجعة الكود البرمجي والتأكد من المعايير
    if "calculate_growth" in code and "ValueError" not in code:
        return False, "الكود يفتقر إلى معالجة استثنائية عند إدخال قيمة صفرية للأساس"
    return True, "تم اعتماد معايير الكود بنجاح"

def qa_agent_run_tests(code: str) -> Tuple[bool, str]:
    # فحص الكود وتشغيله ضد سيناريوهات مختلفة
    local_scope = {}
    exec(code, local_scope)
    fn = local_scope["calculate_growth"]
    
    # فحص حالة الصفر
    try:
        fn(0, 100)
        return False, "فشل الاختبار: إدخال صفر لم يطلق استثناء وقسم على صفر!"
    except ValueError:
        return True, "نجحت جميع اختبارات الوحدة بنسبة 100%"
    except ZeroDivisionError:
        return False, "فشل الاختبار: حدث ZeroDivisionError صريح!"

print("تم تجهيز وكلاء التطوير والمراجعة والفحص.")
```

### 3. تشغيل الحلقة التكرارية وبوابة المشرف البشري (Human-in-the-Loop)

تعال نشغل الدورة المتكاملة:
المطور يكتب -> المراجع يفحص -> الـ QA يختبر -> لو في خطأ يصلحه -> لما كل شيء ينجح يقف النظام ويطلب موافقة البشري قبل النشر!


```python
def run_agile_sprint_cycle(task_id: str, human_approved: bool = True) -> dict:
    print("=" * 60)
    print(f"بدء تنفيذ المهمة البرمجية: {task_id}")
    
    # 1. كود أولي
    code = dev_agent_write_code(task_id)
    
    # 2. مراجعة الكود واكتشاف الملاحظات
    passed_review, rev_note = reviewer_agent_audit(code)
    print(f"[المراجع]: {rev_note}")
    
    # حلقة تصحيح أوتوماتيكية
    if not passed_review:
        code = dev_agent_write_code(task_id, feedback=rev_note)
        passed_review, rev_note = reviewer_agent_audit(code)
        print(f"[المراجع بعد التعديل]: {rev_note}")
        
    # 3. اختبارات الجودة
    passed_qa, qa_note = qa_agent_run_tests(code)
    print(f"[فاحص الجودة QA]: {qa_note}")
    
    # 4. بوابة التدخل البشري (HITL Gate)
    print(f"\n[بوابة المشرف البشري HITL]: فحص النتيجة وطلب الاعتماد...")
    if passed_qa and human_approved:
        deploy_status = "تم النشر للإنتاج بنجاح (DEPLOYED)"
    else:
        deploy_status = "تم تعليق النشر بقرار المشرف البشري (PENDING_HUMAN_REVIEW)"
        
    print(f"القرار النهائي: {deploy_status}")
    print("=" * 60)
    return {"code": code, "status": deploy_status}

# تجربة تشغيل دورة تطوير كاملة مع موافقة البشري
result = run_agile_sprint_cycle("TASK-1", human_approved=True)
```

### 4. جرب بنفسك!

دلوقتي دورك للتجربة الحرة:
1. جرب تشغل الدورة دي بس مع رفض المشرف البشري (`human_approved=False`).
2. لاحظ إزاي النظام، رغم إن كل الاختبارات البرمجية نجحت، بيقف ومينشرش الكود لأن البوابة البشرية لها الكلمة العليا!
3. ده بالضبط مفهوم **Human-in-the-Loop** اللي بيحمي الشركات والمؤسسات من أي قرارات متهورة للذكاء الاصطناعي.


```python
# جرب بنفسك هنا
custom_run = run_agile_sprint_cycle("TASK-1", human_approved=False)

print("ملخص كود الحل المعتمد:")
print(custom_run["code"])
print(f"حالة النشر: {custom_run['status']}")
```
