### 1. تجربة وكيل البنك الفردي (AI Agent)

تعال نبني وكيل خدمة عملاء بسيط لبنك.
الوكيل ده بيعمل مهمتين بس: استعلام الرصيد أو كشف الحساب.
لو طلبت منه أي حاجة تانية برة القواعد المحددة، هيرفض بأدب لأنه وكيل فردي مقيد (Bounded Autonomy).


```python
# قاعدة بيانات حسابات العملاء
accounts_db = {
    "ACC_101": 25000.0,
    "ACC_102": 4500.50
}

def banking_agent(intent: str, account_id: str) -> str:
    # الوكيل بيمشي على قواعد مسبقة التحديد (Predefined Rules)
    if account_id not in accounts_db:
        return f"عفواً، الحساب {account_id} غير مسجل لدينا."
    
    if intent == "رصيد":
        bal = accounts_db[account_id]
        return f"رصيدك الحالي في الحساب {account_id} هو: {bal:,.2f} جنيه."
    elif intent == "كشف_حساب":
        return f"تم تجهيز كشف الحساب لـ {account_id} لآخر 5 معاملات."
    else:
        # رفض أي طلب خارج النطاق الضيق المحدد له
        return f"عفواً، الطلب '{intent}' خارج صلاحيات هذا الوكيل الفردي."

# تعال نجرب الوكيل في حالات مختلفة
print("طلب رصيد:", banking_agent("رصيد", "ACC_101"))
print("طلب قرض:", banking_agent("طلب_قرض", "ACC_101"))
print("حساب غير موجود:", banking_agent("رصيد", "ACC_999"))
```

### 2. تجربة المنظومة الوكالية المتكاملة (Agentic AI)

دلوقتي تعال نشوف الفرق الحقيقي لما نبني منظومة Agentic AI ذكية لإدارة الطاقة في مبنى ذكي.
المنظومة دي بتطبق الأركان الإدراكية الأربعة:
1. **الإدراك (Perception)**: قراءة الحساسات (هل الغرفة فيها ناس؟ درجة الحرارة كام؟ هل دي ساعات ذروة الكهرباء؟).
2. **الاستدلال والتفكير (Reasoning)**: التنسيق بين وكيل الإضاءة، ووكيل التكييف، ووكيل الأجهزة لتحقيق الهدف الكلي (توفير الكهرباء).
3. **التنفيذ (Action)**: إرسال الأوامر الفعلية (إطفاء النور، ضبط التكييف، تأجيل الغسالة).
4. **التعلم والتكيف (Learning)**: مراجعة نسبة التوفير المحققة وتعديل المعاملات تلقائياً للمرات القادمة.


```python
# الوكلاء المتخصصون الفرعيون داخل المنظومة

def lighting_subagent(occupied: bool) -> str:
    # وكيل الإضاءة بيقرر بناءً على التواجد
    return "تشغيل 100%" if occupied else "إطفاء تام 0%"

def thermostat_subagent(occupied: bool, temp_c: float, peak: bool) -> Tuple[str, float]:
    # وكيل التكييف بيوازن بين الراحة وتوفير الكهرباء
    if not occupied:
        return ("وضع الاستعداد", 26.0)
    if peak:
        return ("تبريد اقتصادي في ساعات الذروة", 24.5)
    return ("تبريد مريح طبيعي", 22.0)

def appliance_subagent(peak: bool, current_load_kw: float) -> str:
    # وكيل الأجهزة الثقيلة بيأجل التشغيل لو في ساعات الذروة
    if peak and current_load_kw > 3.0:
        return "تأجيل تشغيل الأجهزة الثقيلة"
    return "تشغيل الأجهزة بشكل طبيعي"

print("تم تجهيز الوكلاء الفرعيين التخصصيين.")
```

### 3. تشغيل الحلقة الإدراكية الرباعية (Perception -> Reasoning -> Action -> Learning)

تعال نشوف العقل المنظومي المشترك وهو بيجمع المعلومات ويدير الوكلاء ويتعلم مع كل دورة.


```python
# العقل المنظومي ونظام التعلم التكيفي
learned_eco_bonus = 0.5  # معامل تعلم يتكيف مع الوقت

def run_agentic_system(occupied: bool, temp_c: float, peak: bool, load_kw: float) -> dict:
    global learned_eco_bonus
    
    # 1. الإدراك (Perception)
    sensor_data = {"occupied": occupied, "temp": temp_c, "peak": peak, "load": load_kw}
    
    # 2. الاستدلال (Reasoning) والتنسيق بين الوكلاء
    light_decision = lighting_subagent(occupied)
    thermo_mode, target_temp = thermostat_subagent(occupied, temp_c, peak)
    appliance_decision = appliance_subagent(peak, load_kw)
    
    # تطبيق التكيف المكتسب من الخبرة
    if peak:
        target_temp += learned_eco_bonus
        
    # 3. التنفيذ (Action)
    actions = {
        "الإضاءة": light_decision,
        "التكييف": f"{thermo_mode} عند {target_temp:.1f} درجة",
        "الأجهزة": appliance_decision
    }
    
    # حساب الاستهلاك التقديري الجديد
    saved_kw = 0.0
    if not occupied:
        saved_kw += 2.0
    if peak:
        saved_kw += 1.2
    new_load = max(0.8, load_kw - saved_kw)
    
    # 4. التعلم والتكيف (Learning)
    saving_ratio = (load_kw - new_load) / load_kw
    # تحديث معامل التكيف التلقائي بناءً على النتيجة
    learned_eco_bonus = round(min(2.5, learned_eco_bonus + 0.1 * saving_ratio), 3)
    
    return {
        "القرارات": actions,
        "الاستهلاك_السابق": f"{load_kw} kW",
        "الاستهلاك_الجديد": f"{new_load:.1f} kW",
        "نسبة_التوفير": f"{saving_ratio * 100:.1f}%",
        "معامل_التعلم_للمرة_القادمة": learned_eco_bonus
    }

# تعال نشغل المنظومة في سيناريو غرفة مشغولة في وقت ذروة
res1 = run_agentic_system(occupied=True, temp_c=35.0, peak=True, load_kw=6.0)
print("الدورة 1 (الغرفة مشغولة في وقت الذروة):")
for k, v in res1.items():
    print(f"  {k}: {v}")

# تعال نشغل دورة تانية بعد خروج السكان من الغرفة
res2 = run_agentic_system(occupied=False, temp_c=33.0, peak=True, load_kw=4.5)
print("\nالدورة 2 (الغرفة خالية - المنظومة تتصرف ذاتياً):")
for k, v in res2.items():
    print(f"  {k}: {v}")
```
