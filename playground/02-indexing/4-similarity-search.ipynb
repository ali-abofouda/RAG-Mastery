```python
import numpy as np

def cosine_similarity(vec1, vec2):
    """
    Calculate the cosine similarity between two vectors.

    Parameters:
    vec1 (numpy.ndarray): First vector.
    vec2 (numpy.ndarray): Second vector.

    Returns:
    float: Cosine similarity between vec1 and vec2.
    """
    dot_product = np.dot(vec1, vec2)
    norm_vec1 = np.linalg.norm(vec1)
    norm_vec2 = np.linalg.norm(vec2)
    
    if norm_vec1 == 0 or norm_vec2 == 0:
        return 0.0  # Return 0 if either vector is zero to avoid division by zero
    
    return dot_product / (norm_vec1 * norm_vec2)
```

```python
docs = [
    "The cat sat on the mat.",  
    "The dog barked at the mailman.",
    "The cat chased the mouse.",
    "The dog and the cat are friends.",
    "The mouse ran away from the cat.",
    "The cat is sleeping on the couch.",
    "The dog is playing with a ball.",
]
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings
import os 
from dotenv import load_dotenv
load_dotenv()
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
embedded_docs = [embeddings.embed_query(doc) for doc in docs]
print("Embedded documents:")
for i, doc in enumerate(docs):
    print(f"Document {i}: {doc}")
    print(f"Embedding: {embedded_docs[i]}\n")
    print(f"Embedding length: {len(embedded_docs[i])}\n")

    print("--------------------------------------------------\n")
```

```python
## clac the cosine similarity between all document and all other documents
similarities = []
for i in range(1, len(embedded_docs)):
    for j in range(i + 1, len(embedded_docs)):
        similarity = cosine_similarity(embedded_docs[i], embedded_docs[j])
        similarities.append((i, j, similarity))
        print(f"Cosine similarity between Document {i} and Document {j}: {similarity}\n")
        print("--------------------------------------------------\n")
```

```python
## example of semantic search using cosine similarity
document_to_search = [
        "langchain is a framework for building applications with LLMs",
        "Ai and machine learning are transforming the world",
        "The future of AI is bright"
]

query = "What is the future of AI?"
```

```python
def semantic_search(query, documents, embeddings):
    """
    Perform semantic search using cosine similarity.

    Parameters:
    query (str): The query string.
    documents (list): List of document strings.
    embeddings (HuggingFaceEmbeddings): Embedding model.

    Returns:
    list: List of tuples containing document index and similarity score.
    """
    query_embedding = embeddings.embed_query(query)
    similarities = []
    
    for i, doc in enumerate(documents):
        doc_embedding = embeddings.embed_query(doc)
        similarity = cosine_similarity(query_embedding, doc_embedding)
        similarities.append((i, similarity))
    
    # Sort by similarity score in descending order
    similarities.sort(key=lambda x: x[1], reverse=True)
    
    return similarities

print("Semantic search results:")
search_results = semantic_search(query, document_to_search, embeddings)
print(search_results) 
# this will print the index of the document and the similarity score expexcted output: [(2, 0.85), (1, 0.75), (0, 0.65)]
```
