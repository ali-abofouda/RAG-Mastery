# Stage 01 - Data Ingestion: Document Data Structure in LangChain
### Lecture 09: Core Document Abstraction, Metadata Schemas, and Serialization

This notebook demonstrates the standard data contract of LangChain's `Document` abstraction:
1. Instantiating `Document` objects with structured payload and metadata.
2. Metadata schema design and field isolation (source, author, timestamps, chunk_id).
3. Serialization to and from native Python dictionaries and JSON payloads.
4. Validation and filtering based on metadata attributes.


```python
# Step 1: Instantiating LangChain Document Objects
from langchain_core.documents import Document

# Define structured document with explicit metadata payload
doc_primary = Document(
    page_content="Retrieval-Augmented Generation combines parametric memory with non-parametric external retrieval.",
    metadata={
        "source": "manual_input",
        "domain": "information_retrieval",
        "author": "AI Research Team",
        "version": 1.0,
        "is_verified": True
    }
)

print(f"[INFO] Instantiated Document:")
print(f"  Content ({len(doc_primary.page_content)} chars): {doc_primary.page_content}")
print(f"  Metadata Attributes: {doc_primary.metadata}")

# Assertions
assert isinstance(doc_primary.page_content, str), "page_content must be of type string."
assert isinstance(doc_primary.metadata, dict), "metadata must be a dictionary."
assert doc_primary.metadata["is_verified"] is True, "Metadata verification flag must be True."
```

```python
# Step 2: Serialization and Deserialization (Data Contract)
import json

# Serialize Document to native dictionary
doc_dict = {
    "page_content": doc_primary.page_content,
    "metadata": doc_primary.metadata
}
json_serialized = json.dumps(doc_dict)
print(f"[INFO] JSON Serialized Payload: {json_serialized}")

# Reconstruct Document from JSON payload
deserialized_data = json.loads(json_serialized)
reconstructed_doc = Document(
    page_content=deserialized_data["page_content"],
    metadata=deserialized_data["metadata"]
)

# Assertions
assert reconstructed_doc.page_content == doc_primary.page_content, "Reconstructed content must match original."
assert reconstructed_doc.metadata["domain"] == doc_primary.metadata["domain"], "Metadata fields must match original."
print("[SUCCESS] Document serialization and reconstruction verified.")
```

```python
# Step 3: Batch Document Operations and Metadata Filtering
docs_batch = [
    Document(page_content="Chunk 1: Overview of Dense Embeddings.", metadata={"chunk_id": 1, "topic": "dense"}),
    Document(page_content="Chunk 2: Overview of BM25 Lexical Inverted Index.", metadata={"chunk_id": 2, "topic": "sparse"}),
    Document(page_content="Chunk 3: Fusion using Reciprocal Rank Fusion.", metadata={"chunk_id": 3, "topic": "hybrid"}),
    Document(page_content="Chunk 4: Cross-Encoder Reranking Architecture.", metadata={"chunk_id": 4, "topic": "rerank"})
]

print(f"[INFO] Batch initialized with {len(docs_batch)} documents.")

# Filter documents matching specific metadata criteria
filtered_docs = [d for d in docs_batch if d.metadata.get("topic") in ["dense", "sparse"]]
print(f"[INFO] Filtered Documents count: {len(filtered_docs)}")
for d in filtered_docs:
    print(f"  [{d.metadata['topic']}]: {d.page_content}")

# Assertions
assert len(docs_batch) == 4, "Batch must contain 4 documents."
assert len(filtered_docs) == 2, "Filtered batch must contain exactly 2 documents."
print("[SUCCESS] Document batch operations and metadata filtering verified.")
```
