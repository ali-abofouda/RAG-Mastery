Demonstrates creating `Document` objects and attaching metadata for vector filtering.

## 1. Import Document Class

```python
from langchain_core.documents import Document

print("✅ Document class imported successfully!")
```

## 2. Create Document with Metadata
Instantiates a Document with `page_content` and domain `metadata` attributes.

```python
# إنشاء كائن Document مع البيانات الوصفية (Metadata)
doc = Document(
    page_content="This is the main text content that will be embedded and searched.",
    metadata={
        "source": "example.txt",
        "page": 1,
        "author": "Krish",
        "date_created": "2024-01-01",
        "department": "Engineering"
    }
)

print(f"Document Type: {type(doc)}")
```

## 3. Inspect Content & Metadata Fields

```python
# طباعة المحتوى النصي
print("--- Page Content ---")
print(doc.page_content)

# طباعة الميتا-داتا بالكامل
print("\n--- Metadata Dictionary ---")
print(doc.metadata)

# الوصول لحقول محددة داخل الميتا-داتا
print(f"\nSource File: {doc.metadata['source']}")
print(f"Page Number: {doc.metadata['page']}")
print(f"Author: {doc.metadata['author']}")
```

## 4. Metadata Filtering Example
Demonstrates how metadata enables post-retrieval filtering in vector stores.

```python
# إنشاء قائمة مستندات مع ميتا-داتا متنوعة
docs = [
    Document(page_content="تقرير المبيعات للربع الأول 2024", metadata={"dept": "Sales", "year": 2024}),
    Document(page_content="سياسة الإجازات السنوية للموظفين", metadata={"dept": "HR", "year": 2024}),
    Document(page_content="معايير الأمان للبيانات لعام 2023", metadata={"dept": "IT", "year": 2023}),
    Document(page_content="دليل الموظف الجديد والمكافآت", metadata={"dept": "HR", "year": 2024}),
]

# فلترة المستندات الخاصة بقسم الموارد البشرية لعام 2024 فقط
hr_2024_docs = [
    d for d in docs 
    if d.metadata.get("dept") == "HR" and d.metadata.get("year") == 2024
]

print(f"عدد الوثائق المطابقة للفلترة: {len(hr_2024_docs)}")
for idx, d in enumerate(hr_2024_docs, 1):
    print(f"{idx}. {d.page_content} (Dept: {d.metadata['dept']}, Year: {d.metadata['year']})")
```
