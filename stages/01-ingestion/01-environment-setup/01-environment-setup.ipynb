# Stage 01 - Data Ingestion: Environment and Runtime Setup
### Lecture 08: Virtual Environment & Core Dependency Verification

This notebook establishes deterministic runtime environment verification:
1. Python version compatibility check (Python >= 3.10).
2. Verification of active virtual environment.
3. Import verification for core GenAI and RAG libraries (LangChain, Pydantic, NumPy).
4. System environment variables and data path presence check.


```python
# Step 1: Python Runtime and Virtual Environment Audit
import sys
import os
from pathlib import Path

# Verify Python major and minor version
python_version = sys.version_info
print(f"[CONFIG] Python Version: {python_version.major}.{python_version.minor}.{python_version.micro}")

# Verify active virtual environment
in_virtual_env = sys.prefix != sys.base_prefix
print(f"[CONFIG] Virtual Environment Active: {in_virtual_env} (Prefix: {sys.prefix})")

# Assertions
assert python_version.major == 3 and python_version.minor >= 10, "Python 3.10+ is strictly required."
assert in_virtual_env, "Execution must run within a dedicated virtual environment."
```

```python
# Step 2: Core Dependency Imports and Version Logging
import langchain
import langchain_core
import langchain_community
import pydantic
import numpy as np
from dotenv import load_dotenv

# Load local environment configuration
load_dotenv()

print(f"[INFO] LangChain Version: {langchain.__version__}")
print(f"[INFO] LangChain Core Version: {langchain_core.__version__}")
print(f"[INFO] Pydantic Version: {pydantic.__version__}")
print(f"[INFO] NumPy Version: {np.__version__}")

# Assertions
assert hasattr(langchain, "__version__"), "LangChain package must be properly installed."
assert hasattr(pydantic, "__version__"), "Pydantic package must be properly installed."
```

```python
# Step 3: Repository Root and Data Path Verification
# Locate repository root deterministically
repo_root = Path.cwd()
while not (repo_root / "data").exists() and repo_root.parent != repo_root:
    repo_root = repo_root.parent

data_dir = repo_root / "data"
print(f"[INFO] Detected Repository Root: {repo_root}")
print(f"[INFO] Verified Data Directory: {data_dir} (Exists: {data_dir.exists()})")

# Assertions
assert data_dir.exists() and data_dir.is_dir(), "Standard 'data/' directory must exist in repository root."
print("[SUCCESS] Environment and dependency verification completed successfully.")
```
