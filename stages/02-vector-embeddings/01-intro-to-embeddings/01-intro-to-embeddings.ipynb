# Stage 02 - Vector Embeddings: Fundamentals of Dense Vector Representation
### Lecture 19: Token-to-Vector Mapping and Semantic Space Foundations

This notebook establishes the foundational mechanics of text embeddings:
1. Transforming discrete textual tokens into dense, continuous multidimensional vector spaces.
2. Inspecting vector dimensions, magnitude, and numerical properties.
3. Basic vector distance metrics (Euclidean L2 vs Cosine Distance).
4. Demonstrating semantic clustering in high-dimensional space.


```python
# Step 1: Environment Setup and Model Initialization
import numpy as np
from sentence_transformers import SentenceTransformer

# Load a lightweight, robust embedding model
model_name = "all-MiniLM-L6-v2"
embedding_model = SentenceTransformer(model_name)

print(f"[CONFIG] Embedding Model: {model_name}")
print(f"[CONFIG] Target Embedding Dimension: {embedding_model.get_sentence_embedding_dimension()}")

# Assertion Checkpoint
assert embedding_model.get_sentence_embedding_dimension() == 384, "MiniLM model must output 384-dimensional vectors."
```

```python
# Step 2: Generating and Inspecting Dense Vectors
sample_sentences = [
    "Artificial intelligence transforms modern computing systems.",
    "Machine learning models learn patterns from training data.",
    "The chef prepared a delicious seafood dish with garlic butter."
]

embeddings = embedding_model.encode(sample_sentences)
print(f"[INFO] Generated embeddings matrix shape: {embeddings.shape}")

for idx, (sent, vec) in enumerate(zip(sample_sentences, embeddings), start=1):
    vector_norm = np.linalg.norm(vec)
    print(f"  Sentence {idx} ({sent[:45]}...):")
    print(f"    Vector Shape: {vec.shape} | L2 Norm: {vector_norm:.4f}")
    print(f"    Sample Dimensions (first 5 values): {np.round(vec[:5], 4)}")

# Assertions
assert embeddings.shape == (3, 384), "Embeddings matrix must match (number_of_sentences, dimension)."
assert all(np.isfinite(vec).all() for vec in embeddings), "Vectors must contain valid finite numbers."
```

```python
# Step 3: Pairwise Distance and Semantic Clustering
from sklearn.metrics.pairwise import cosine_similarity, euclidean_distances

# Compute Cosine Similarity Matrix (1.0 = identical direction, 0.0 = orthogonal)
cos_matrix = cosine_similarity(embeddings)
l2_matrix = euclidean_distances(embeddings)

print("[INFO] Pairwise Cosine Similarity Matrix:")
print(np.round(cos_matrix, 4))

sim_ai_ml = cos_matrix[0, 1]     # AI vs ML (Related)
sim_ai_food = cos_matrix[0, 2]   # AI vs Food (Unrelated)

print(f"[METRICS] AI vs ML Similarity: {sim_ai_ml:.4f}")
print(f"[METRICS] AI vs Food Similarity: {sim_ai_food:.4f}")

# Assertions
assert sim_ai_ml > sim_ai_food, "Semantically related sentences (AI vs ML) must have higher similarity than unrelated (AI vs Food)."
assert np.isclose(cos_matrix[0, 0], 1.0), "Self-similarity must equal 1.0."
print("[SUCCESS] Vector embedding fundamentals verified.")
```
